## Funciones
### Una funcion agrupa un codigo reutilizable

```python
def clasificar_edad(edad):
    """Recibe una edad o regresa la categoria correspondiente"""
    if edad < 12:
        return "Niño"
    elif edad < 18:
        return "Adolescente"
    elif edad < 65:
        return "Adulto"
    else:
        return "Adulto mayor"

# probamos la función con distintos valores
for edad in [5, 15, 40, 82]:
    print(edad, "->", clasificar_edad(edad))
```

```python
lista_pasajeros = []
contador = 0   # empezamos con 0 pasajeros

def agregar_pasajero():
    global contador
    if contador < 4:   # máximo 4 pasajeros
        nombre = input("Nombre del pasajero: ")
        edad = input("Edad: ")
        clase = input("Clase: ")

        pasajero = {
            "nombre": nombre,
            "edad": edad,
            "clase": clase
        }

        lista_pasajeros.append(pasajero)
        contador += 1
    else:
        print("No se pueden agregar más de 4 pasajeros")

# Probamos
for i in range(4):
    agregar_pasajero()

print("\nLista de pasajeros:")
for p in lista_pasajeros:
    print(p)

    
```
